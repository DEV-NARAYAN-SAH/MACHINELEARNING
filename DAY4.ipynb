DataFrame

A DataFrame is a 2-dimensional, size-mutable, tabular data structure with labeled axes (rows and columns).
Think of it like an Excel sheet or SQL table.

```python
import pandas as pd 
data={
    "name":["bill","tom","tim","dev","alex","vanessa","kate"],
    "score":[90,80,85,75,95,60,65],
    "sport":["wrestling","football","skiing","swimming","tennis","karete","surfing"],
    "sex":["m","m","m","m","m","f","f"]
}
df=pd.DataFrame(data)
```

Accessing and indexing data in dataframe

```python
df
```

```python
#Accesing a single column
name_data_frame = df ['name']

##accessing multiple columns
print(df[['name','sport','score']])
```

```python
## Accessing a specific row by index
print(df.iloc[2]) #third row (index 2)

##accessing a range of rows using slicing
print(df.iloc[1:4])# rows with indices 1,2,3
```

```python
## Accessing a specific row by index
print(df.iloc[2]) #third row (index 2)

##accessing a range of rows using slicing
print(df.loc[1:4,['name','score']])# rows with indices 1,2,3
```

indexing with .loc[] and .iloc[]

- .loc[]:label:based indexing.it works with labels(row/column names)
- .iloc[]:position -based indexing. it works with integer positions(row/colum indices)

slicing

- with .loc[]: both the start and stop are include.
- with .iloc[]: the stop is excluded.


```python
df
```

```python
#using .loc for slicing (inclusive of end )
print(df.loc['bill':'tim','score':'sex'])
```

```python
#using .loc for slicing (inclusive of end )
print(df.loc['bill':'tim','score':'sport'])
```

```python
#using .iloc for slicing(exclusive of end)
print(df.iloc[1:3,0:2])
```

Filtering Data (boolean indexing)

```python
# filtering based on a condition(e.g, score>80)
df[df['sex']=='m']
```

```python
# filtering based on multiple conditions
print(df[(df['score']>80) & (df['sex']=='m')])
```

sorting data

```python
#sorting by score
print(df.sort_values('score', ascending=False))
```

```python
print(df.sort_values(by=['sex','score'], ascending=[True,False]))
```
