combinging and merging datasets in pandas

```python
import pandas as pd
```

joining Dataframe 

```python
d1=pd.DataFrame(
    {"key":['a','b','c','d','e','f'],
    "num1":range(6)}
)

d2=pd.DataFrame(
    {"key":['b','c','e','f'],
    "num2":range(4)}
)
```

```python
print(d1)
print(d2)
```

```python
pd.merge(d1,d2)
```

```python
pd.merge(d1,d2,on='key')
```

dataset having different key names

```python
d3=pd.DataFrame(
    {"key1":['a','b','c','d','e','f'],
    "num1":range(6)}
)

d4=pd.DataFrame(
    {"key2":['b','c','e','f'],
    "num2":range(4)}
)
```

```python
print(d3)
print(d4)
```

```python
pd.merge(
    d3,d4,left_on='key1',right_on="key2")
```

outer join

all rows from both datframes,with NAN where no match is found

```python
print(d1)
print(d2)
```

```python
pd.merge(d1,d2,how='outer')
```

left join

```python
pd.merge(d1,d2,how='left')
```

right join

```python
pd.merge(d1,d2,how='right')
```

merging on index

```python
df1=pd.DataFrame(
    {"letter":['a','a','b','b','a','c'],
     'num':range(6)
    
})
df2=pd.DataFrame(
    {"value":[3,5,7]},
     index=['a','b','e']
    
)
```

```python
print(df1)
print(df2)
```

```python
pd.merge(df1,df2,
        left_on='letter',
        right_index=True)
```

```python
data1=pd.DataFrame(
    [[1,2],[3,4],[5,6]],
    index=['a','c','d'],
    columns=['tom','tim']
)

data2=pd.DataFrame(
    [[7,8],[9,10],[11,12],[13,14]],
    index=['a','b','e','f'],
    columns=['sam','kim']
)
```

```python
print(data1)
print(data2)
```

```python
pd.merge(data1,data2,
        right_index=True,
        left_index=True,
        how='outer' )
```

```python
data2.join(data1)
```

```python
data1.join(data2,how='outer')
```
