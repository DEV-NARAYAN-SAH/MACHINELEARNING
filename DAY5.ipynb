GroupBY

groupby() function in pandas is used to splite the data into groups based on some criteria
and we can apply a fuction to each group independently

```python
data = {'category':['A','B','A','B','A'],
       'values':[10,20,30,40,50]}
df = pd.DataFrame(data)
print(df)

#group by 'category' and calculated the sum of 'values'
grouped = df.groupby('category')

grouped.sum()
```

```python
grouped.mean()
```

Missing Data in Pandas

```python
import pandas as pd 
import numpy as np
```

```python
from numpy import nan as NA
```

```python
df = pd.DataFrame([[1,2,3],[4,None,5],
                 [NA,NA,NA]])
df
```

checking NaN

```python
df.isnull().any()
```

droprow or column having Na

```python
df.dropna(inplace=False)
```

```python
df
```

```python
df.dropna(how='all')
```

```python
df
```

```python
df=pd.DataFrame([[1,2,3],[4,NA,5],
                 [NA,NA,NA]])
df
```

```python
df.dropna(axis=1, how='all')
```

```python
df
```

Fill NaN values with other values

```python
df.fillna(0)
```

```python
df.fillna({0:15,1:25,2:35}, inplace=True)
```

```python
df
```

with inplace- modifies the calling object

```python
df.fillna(0,inplace=True)
df
```

```python
forward fill-fill method propagates the last valid (non-null) values forward to replace any missing values
axis=0:fill
axis=1:fill row-wise
```

```python
df
```

```python
df.fillna(method='ffill',axis=1)
#df.ffill(axix=1)
```

```python
df
```

```python
df.fillna(method='ffill',limit=1)
```

```python
df
```

```python
df.fillna(df.mean(axis=1))
```

```python
df.fillna(df.mean(axis=1))
```
