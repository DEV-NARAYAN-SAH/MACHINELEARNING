Dictionary to series

```python
# creating using dictionary

score={"jane":90, "bill":80, "elon":85, "tom":75, "tim":95, "gp":None}
names=pd.Series(score)
names
```

```python
names['tim']
```

Boolean Indexing

filter elements using conditional logic:

```python
names[names>=85]
```

Modifying values
- updates values using labels or conditions: 

```python
names['tom']=60
names[names <= 80]=83
names
```

Checking memberships
- verify if a label exists in the series:

```python
"tom" in names
```

```python
if 'tom' in names:
    print('True')

else:
    print('False')
```

```python
'dev' in names
```

```python
names['dev']=20
```

```python
if 'dev' in names:
    print('whats up')

else:
    print('hello')
```
