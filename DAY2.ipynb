BROADCASTING

```python
import numpy as np

a =  np.array([1,2,3]) 
print(a)

b = np.array([[4],[5],[6]])
print(b)

c=a+b
print(c)
```

```python
import numpy as np

a =  np.array([[1],[2],[3]]) # has shape(3,1)
print(a)

b = np.array([[4,5],[6,7],[8,9]]) # has shape(3,2)
print(b)

c=a+b                             # will have shape(3,2)
print(c)
```

```python
import numpy as np

a =  np.array([[1],[2],[3],[4]]) # has shape(4,1)
print(a)

b = np.array([[4,5],[6,7],[8,9]]) # has shape(3,2)
print(b)

c=a+b                             # will have shape(3,2)
print(c)
```

```python
import numpy as np

a =  np.array([[1],[2],[3],[4]]) # has shape(4,1)
print(a)

b = np.array([[4,5],[6,7],[8,9],[10,11]]) # has shape(4,2)
print(b)

c=a+b                             # will have shape(4,2)
print(c)
```

 LINEAR ALGEBRA WITH NUMPY

```python
import numpy as np
a = np.array([[1,2],[3,4]])

#Determine of a matrix
det_a = np.linalg.det(a)
print('Determine of a:',det_a)
```

```python
import numpy as np

a = np.array([[1,2,3,4],[3,4,5,6],[6,7,8,9],[2,3,4,5]])
print(a)

det_a = np.linalg.det(a)
print('Determine of a:',det_a)
```

```python
import numpy as np

a = np.array([[13,24,35,46],[3,4,5,6],[6,7,8,9],[12,13,14,15]])
print(a)

det_a = np.linalg.det(a)
print('Determine of a:',det_a)
```

```python
# Example system of linear equations:
# 2x+y=5
# x-y=1

a = np.array([[2,1],[1,-1]])
b = np.array([5,1])
print(a)
print(b)

solutions = np.linalg.solve(a,b)
print('solution:',solutions)
```

```python
a = np.array([[3,1,-2],[2,-1,1],[1,2,3]])
b = np.array([9,3,12])
print(a)
print(b)
solutions = np.linalg.solve(a,b)
print("solution:",solutions)
```

RANDOM NUMBER WITH NUMPY

```python
import numpy as np

# generate a single random number between 0 and 1
random_number = np.random.rand()
print(random_number)

# generate a 1D array of 5 random number
random_array_1d = np.random.rand(5)
print(random_array_1d)

random_array_2d = np.random.rand(3,3)
print(random_array_2d)
```

```python
#generate  a single random integer between 0 and 10
random_int = np.random.randint(0,10)
print(random_int)

# generate a 1d array of 5 random integers between 0 and 10
random_int_array = np.random.randint(0,10, size=5)
print(random_int_array)

# generate a 2d array of 3*2 matrix random integers between 0 and 10
random_int_array_2d = np.random.randint(0,10, size=(3,2))
print(random_int_array_2d)
```

Random samples from a given list

```python
random_sample = np.random.choice([1,2,3,4,5], size=3)
print(random_sample)

random_sample_with_replacement = np.random.choice([1,2,3,4,5], size=3, replace=True)
print(random_sample_with_replacement)
```

Random numbers with a specific seed (reproductibility)

```python
# Set the seed for reproducibility
import numpy as np
np.random.seed(42)

# Generate random numbers
random_numbers = np.random.rand(3)
print(random_numbers)  # Corrected variable name

# Set the seed again and generate the same numbers
np.random.seed(42)
random_numbers_again = np.random.rand(3)
print(random_numbers_again)
```
