```python
# Add 
import numpy as np
x = np.array([[1,2],[3,4]], dtype= np.float64)
y = np.array([[5,6],[7,8]],dtype= np.float64)
#elementwise sum; both produce the array
print(x+y)
print(np.add(x,y))
```

```python
#elementwise subtract; both produce the array
print(x-y)
print(np.subtract(x,y))
```

```python

#elementwise multiplication; both produce the array
print(x*y)
print(np.multiply(x,y))
```

```python
#elementwise division; both produce the array
print(x/y)
print(np.divide(x,y))
```

```python
#elementwise square root; both produce the array

print(np.sqrt(x))
```

```python
import numpy as np

x = np.array([[1,2],[3,4]])
y = np.array([[5,6],[7,8]])

v = np.array([9,10])
w = np.array([11,12])
# inner product of vectors; both produce 219
print(v.dot(w))
print(np.dot(v ,w))
```

```python
print(x.dot(v))
print(np.dot(x,v))
print(x@v)
```

```python
import numpy as np 

# define the features of the house 
features = np.array([1500, 3, 10]) # [square footage ,number of bedrooms, age of the house]

#design the weights (importance of each feature)
weights = np.array([100, 5000, -2000]) # weights corresponding to [square footage, bedrooms , age]

#calculate the predicted price using the dot product
predicted_price = np.dot(features,weights)

print(f"The predicted price of the house is: ${predicted_price}" )
```

```python
a = np.array([1,8,9,-3,2,4,7,9])
# get the index of the maximum element in a
print(np.argmax(a))
```
